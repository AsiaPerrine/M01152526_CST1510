### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
#D6
"""Check the status of a percentage."""
def status_of(percent, warning_at=90):
    """Return the status based on the percentage."""

    # Check the percentage and return the status
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

# Call the function with different percentages
print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

### D12. Returning more than one value

1. Define `check_record(value, limit)` that works out `difference = value - limit` and
   `percent = (value / limit) * 100`, and returns **both**: `return difference, percent`

3. Print them.

**Expected output:**
```

```

````python
"""Calculate the difference and percentage of a record."""
def check_record(value, limit):
    """Return the difference and percentage."""

    # Calculate the difference and percentage
    difference = value - limit
    percent = (value / limit) * 100

    # Return both results
    return (difference, percent)

# Call the function and store the results
difference, percent = check_record(87, 100)

# Display the results
print(f"Difference: {difference:.1f}")
print(f"Percent: {percent}")
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
# C1
"""Check the status of different records."""
def percent_of(value, limit):
    """Calculate the percentage of a value."""

    # Calculate and return the percentage
    return (value / limit) * 100


def status_of(percent):
    """Return the status based on the percentage."""

    # Check if the percentage is over the limit
    if percent >= 100:
        return "OVER LIMIT"
    else:
        return "OK"

# Call both functions and display the results
print(status_of(percent_of(87, 100)))
print(status_of(percent_of(120, 100)))
print(status_of(percent_of(45, 100)))
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
"""Count how many values are over the limit."""
def count_over(*values, limit=100):
    """Count how many values are over the limit."""
    # Start the counter at 0
    count = 0

    # Check each value against the limit
    for val in values:
        if val > limit:
            count += 1

    # Return the total count
    return count

# Count values using the default limit of 100
print(count_over(87, 120, 45, 101))

# Count values using a limit of 50
print(count_over(87, 120, 45, 101, limit=50))

# Call the function without any values
print(count_over())
````
